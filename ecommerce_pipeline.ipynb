```python
# ===== CLEAN THE UNCLEAN EXCEL FILE (fixed pipeline) =====
import pandas as pd
import numpy as np

# 1. Always re-read the unclean file fresh (this picks up any rows you just added)
df = pd.read_excel("Ecommerce_Unclean_Project.xlsx")
print("Rows loaded from unclean file:", len(df))

# 2. Clean column names
df.columns = df.columns.str.strip()

# 3. Strip whitespace from all text columns, normalize blank-like values to NaN
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].astype(str).str.strip()
    df[col] = df[col].replace({"nan": np.nan, "": np.nan, "None": np.nan, "NaT": np.nan})

# 4. Drop fully duplicate rows
df = df.drop_duplicates()

# 5. Drop duplicate Order_ID, keep the first occurrence
if "Order_ID" in df.columns:
    df = df.drop_duplicates(subset="Order_ID", keep="first")

# 6. Drop rows missing an essential identifier
essential_cols = [c for c in ["Order_ID", "Order_Date"] if c in df.columns]
if essential_cols:
    df = df.dropna(subset=essential_cols)

# 7. Parse dates (dayfirst since your data uses dd-mm-yyyy)
for date_col in ["Order_Date", "Delivery_Date"]:
    if date_col in df.columns:
        df[date_col] = pd.to_datetime(df[date_col], errors="coerce", dayfirst=True)

if "Order_Date" in df.columns:
    df = df.dropna(subset=["Order_Date"])

# 8. Numeric cleanup
for num_col in ["Qty", "Unit_Price", "Discount"]:
    if num_col in df.columns:
        df[num_col] = pd.to_numeric(df[num_col], errors="coerce")

needed_numeric = [c for c in ["Qty", "Unit_Price"] if c in df.columns]
if needed_numeric:
    df = df.dropna(subset=needed_numeric)

# 9. Derive Sales / Net_Amount / Profit if the unclean file doesn't already have them
if "Sales" not in df.columns and {"Qty", "Unit_Price"}.issubset(df.columns):
    df["Sales"] = df["Qty"] * df["Unit_Price"]

if "Net_Amount" not in df.columns and {"Sales", "Discount"}.issubset(df.columns):
    df["Net_Amount"] = df["Sales"] * (1 - df["Discount"].fillna(0) / 100)

# NOTE: 20% is a PLACEHOLDER profit margin — replace with your real profit formula/column
if "Profit" not in df.columns and "Net_Amount" in df.columns:
    df["Profit"] = df["Net_Amount"] * 0.20

# 10. Derive Month / Year / Weekday from Order_Date, then format dates back to dd-mm-yyyy strings
if "Order_Date" in df.columns:
    df["Month"] = df["Order_Date"].dt.month_name()
    df["Year"] = df["Order_Date"].dt.year
    df["Weekday"] = df["Order_Date"].dt.day_name()
    df["Order_Date"] = df["Order_Date"].dt.strftime("%d-%m-%Y")
    if "Delivery_Date" in df.columns:
        df["Delivery_Date"] = df["Delivery_Date"].dt.strftime("%d-%m-%Y")

print("Rows after cleaning:", len(df))
print("Columns:", list(df.columns))
df.head()
```

```python
# ===== PUSH CLEANED DATA TO THE SQL TABLE POWER BI READS FROM =====
# This is the step that was MISSING before — without it, Power BI never sees your new rows.

# Save a clean CSV too (optional, in case you ever want to point Power BI at the file directly)
df.to_csv("Clean_Ecommerce.csv", index=False)

df.to_sql(
    "ecommerce_data",
    con=engine,
    if_exists="replace",
    index=False
)

print("Cleaned data pushed to SQL. Rows now in ecommerce_data table:", len(df))
```
